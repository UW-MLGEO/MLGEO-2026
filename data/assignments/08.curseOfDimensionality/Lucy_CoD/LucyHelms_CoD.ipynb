Curse of Dimensionality assignment

```python
# load dependencies
import numpy as np
import matplotlib.pyplot as plt
```

```python
# generate 1000 normally distributed (x,x) data points with a mean of 0 and standard deviation of 0
rng = np.random.default_rng() 
x_int = rng.normal(0, 1, size=1000)
y_int = rng.normal(0, 1, size=1000)
coord_int = (x_int, y_int)

# calculate euclidean distances of all points from (1.5,1.5)
euc_dist = np.sqrt((x_int - 1.5)**2 + (y_int - 1.5)**2)
within_std = euc_dist <= 1  # within one standard deviation
```

```python
# plots!

# 1. Create a figure and a set of subplots
# multi panel figure with 2 rows and 2 columns
fig, axs = plt.subplots(2, 2)

#create a scatterplot of random (x,y) values
axs[0,0].scatter(x_int[~within_std], y_int[~within_std], label = "outside s.d. of (1.5, 1.5)")
axs[0,0].scatter(x_int[within_std], y_int[within_std], c="red", label="within s.d. of (1.5, 1.5)")
axs[0,0].set_xlabel("Random X coordinate")
axs[0,0].set_ylabel("Random Y coordinate")
axs[0,0].set_title("X,Y")
axs[0,0].legend()

#creat histogram of X coordinate
axs[0,1].hist(x_int[~within_std], bins = 25, label = "outside s.d. of (1.5, 1.5)")
axs[0,1].hist(x_int[within_std], bins = 25, color="red", label="within s.d. of (1.5, 1.5)")
axs[0,1].set_xlabel("Random X coordinate")
axs[0,1].set_ylabel("Frequency")
axs[0,1].set_title("Histogram of X")
axs[0,1].legend()

#create a histogram of y coordinate
axs[1,0].hist(y_int[~within_std], bins = 25, label = "outside s.d. of (1.5, 1.5)")
axs[1,0].hist(y_int[within_std], bins = 25, color="red", label="within s.d. of (1.5, 1.5)")
axs[1,0].set_xlabel("Random Y coordinate")
axs[1,0].set_ylabel("Frequency")
axs[1,0].set_title("Histogram of Y")
axs[1,0].legend()

axs[1,1].text(0.05 , 0.3, "Less data is captured within \n the same range in 2D \n (top left), when compared \n with the same range in 1D \n (top right, bottom left)")
axs[1,1].tick_params(axis="both", colors="white")
plt.tight_layout()

```

```python
# percentage of X,Y coordinates within 1 s.d. of (1.5,1.5)
sd_scatter = np.sum(within_std) / len(coord_int)
print(sd_scatter)

# percentage of X values within 1 s.d. of 1.5
sd_x = np.sum(within_std) / len(x_int)
print(sd_x)

# percentage of Y values within 1 s.d. of 1.5
sd_y = np.sum(within_std) / len(y_int)
print(sd_y)
```
