## The Curse of Dimensionality Assignment 

```python
import pandas
import numpy as np 
import matplotlib.pyplot as plt
```

```python

# Parameters
dims = np.arange(2, 501, 5)
trials = 1000

means = []
stds = []
mins = []
maxs = []

for d in dims:
    corrs = []
    
    for _ in range(trials):
        x = np.random.randn(d)
        y = np.random.randn(d)
        
        rho = np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))
        corrs.append(rho)
        
    corrs = np.array(corrs)
    
    means.append(np.mean(corrs))
    stds.append(np.std(corrs))
    mins.append(np.min(corrs))
    maxs.append(np.max(corrs))

means = np.array(means)
stds = np.array(stds)
mins = np.array(mins)
maxs = np.array(maxs)
```

```python
plt.figure(figsize=(10,6))

# Mean line
plt.plot(dims, means, label="Mean", linewidth=2)

# ±1 SD shading
plt.fill_between(dims,
                 means - stds,
                 means + stds,
                 alpha=0.3,
                 label="±1 SD")

# Min and max lines
plt.plot(dims, mins, linestyle="--", label="Min")
plt.plot(dims, maxs, linestyle="--", label="Max")

plt.xlabel("Dimension (d)")
plt.ylabel("Correlation")
plt.title("Concentration of Correlation as Dimension Increases")
plt.legend()
plt.grid(True)
plt.show()
```
