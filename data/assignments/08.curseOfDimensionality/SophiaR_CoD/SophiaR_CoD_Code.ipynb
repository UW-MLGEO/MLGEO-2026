## <u>Dependencies</u>

```python
import numpy as np

import matplotlib.pyplot as plt
```

## <u>Scatter Plot Generation</u>


```python
# Create 1000 random points normally distributed with mean 0 and std 1
np.random.seed(42)
points = np.random.normal(0, 1, (1000, 2))
x, y = points[:, 0], points[:, 1]

# Center point and standard deviation radius
center_x, center_y = 1.5, 1.5
std = 1

# Calculate distance from point (1.5, 1.5)
distances = np.sqrt((x - 1.5)**2 + (y - 1.5)**2)
within_std = distances <= 1  # within one standard deviation
```

```python
# Create scatter plot figure
fig, ax1 = plt.subplots(figsize=(7, 5))  # Use a single axis

ax1.scatter(x[~within_std], y[~within_std], c='black', alpha=0.6, label='Beyond 1σ')
ax1.scatter(x[within_std], y[within_std], c='green', alpha=0.6, label='Within 1σ of (1.5, 1.5)')

# Add vertical boundary lines
ax1.axvline(center_x - std, color='green', linestyle='--', alpha=0.7)
ax1.axvline(center_x + std, color='green', linestyle='--', alpha=0.7)

# Add horizontal boundary lines
ax1.axhline(center_y - std, color='green', linestyle='--', alpha=0.7)
ax1.axhline(center_y + std, color='green', linestyle='--', alpha=0.7)


ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('Randomly Generated Scatter Plot')
ax1.legend()
ax1.grid(True, alpha=0.3)

plt.show()
```

## <u>Histogram Generation</u>

```python
# Create a figure and axis for the histogram
fig, ax2 = plt.subplots(figsize=(7, 5))  # Use a single axis

# Create histogram for both x and y values
ax2.hist(x[~within_std], bins=30, color='black', alpha=0.6, label='Beyond 1σ')
ax2.hist(x[within_std], bins=30, color='green', alpha=0.6, label='Within 1σ of 1.5')
ax2.hist(y[~within_std], bins=30, color='black', alpha=0.3)
ax2.hist(y[within_std], bins=30, color='green', alpha=0.3)
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')
ax2.set_title('Histogram of X and Y')
ax2.legend()
ax2.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

### <u>Final Plots</u>

```python
# Create a figure with two subplots side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Scatter plot on the first axis
ax1.scatter(x[~within_std], y[~within_std], c='black', alpha=0.6, label='Beyond 1σ')
ax1.scatter(x[within_std], y[within_std], c='green', alpha=0.6, label='Within 1σ of (1.5, 1.5)')
# Add vertical boundary lines
ax1.axvline(center_x - std, color='green', linestyle='--', alpha=0.7)
ax1.axvline(center_x + std, color='green', linestyle='--', alpha=0.7)

# Add horizontal boundary lines
ax1.axhline(center_y - std, color='green', linestyle='--', alpha=0.7)
ax1.axhline(center_y + std, color='green', linestyle='--', alpha=0.7)

ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('Randomly Generated Scatter Plot')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Histogram on the second axis
ax2.hist(x[~within_std], bins=30, color='black', alpha=0.6, label='Outside 1σ')
ax2.hist(x[within_std], bins=30, color='green', alpha=0.6, label='Within 1σ of 1.5')
ax2.hist(y[~within_std], bins=30, color='black', alpha=0.3)
ax2.hist(y[within_std], bins=30, color='green', alpha=0.3)
ax2.set_xlabel('Value')
ax2.set_ylabel('Frequency')
ax2.set_title('Histogram of X and Y Values')
ax2.legend()
ax2.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

### <u>Data Interpretation</u>
Here we can analyze the percentages of 1D and 2D values within 1 SD of 1.5

```python
# 1. % of 2D points within 1 std of (1.5, 1.5)
percent_scatter = np.sum(within_std) / len(points) * 100

# 2. % of 1D values within 1 std of 1.5
all_values = np.concatenate([x, y])
within_1d = np.abs(all_values - 1.5) <= 1
percent_hist = np.sum(within_1d) / len(all_values) * 100

# Results: 
print(f"Scatter points within 1σ of (1.5, 1.5): {percent_scatter:.2f}%")
print(f"Histogram values within 1σ of 1.5: {percent_hist:.2f}%")
```
