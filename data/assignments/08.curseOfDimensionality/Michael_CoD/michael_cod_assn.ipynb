# The Curse of Dimensionality
We will plot data in increasing dimensions to reveal how data the distance between points converge.

```python
# Dependencies
import numpy as np
import math
import matplotlib.pyplot as plt
import random as r
from scipy.spatial.distance import pdist
```

```python
# Define dimensions 1 through 15
dimensions = list(range(1, 16))

# Get mean and std dev from first dimension
num_points = 1000
points_1d = np.random.rand(num_points, 1)
distances_1d = pdist(points_1d)
mean_distance_1d = np.mean(distances_1d)
std_dev_1d = np.std(distances_1d)

# Plot dimension vs. percent of pairwise distances within one std dev of the first dimension
percents_within_1d_std = []
for dim in dimensions:
    num_points = 1000
    points = np.random.rand(num_points, dim)
    distances = pdist(points)
    within_std = np.sum((distances >= mean_distance_1d - std_dev_1d) & (distances <= mean_distance_1d + std_dev_1d))
    percent_within_std = (within_std / len(distances)) * 100
    percents_within_1d_std.append(percent_within_std)

plt.figure()
plt.plot(dimensions, percents_within_1d_std, marker='o')
plt.title('Percent of Pairwise Distances Within One Std Dev of 1D Mean by Dimension')
plt.xlabel('Dimension')
plt.ylabel('Percent Within One Std Dev of 1D Mean')
plt.grid()
plt.savefig('michael_cod_fig.png')
plt.show()
```
