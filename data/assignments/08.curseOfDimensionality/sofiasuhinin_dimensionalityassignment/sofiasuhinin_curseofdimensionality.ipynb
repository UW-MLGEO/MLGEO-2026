```python
# import dependencies

import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist
```

```python
#create a set of random points to illustrate the curse of dimensionality
np.random.seed(67)

dimensions = np.arange(1, 201, 5)
n_points = 500

percentages = [] #create list to later append 

for d in dimensions:
    points = np.random.rand(n_points, d)      
    distance = pdist(points) #calculate distances between points

    mean = np.mean(distance) #calculate mean distance
    std  = np.std(distance) #calculate standard deviation of distances

    within_std = np.abs(distance - mean) <= std 
    percent_within_std = np.mean(within_std) * 100

    percentages.append(percent_within_std)


x, y = points[:, 0], points[:, 1]

plt.figure(figsize=(8, 6))
plt.scatter(x, y, alpha=0.5)
plt.title("Scatter Plot of Random Point Generation")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()


plt.figure(figsize=(8,6))
plt.plot(dimensions, percentages, linewidth=2)
plt.xlabel("Dimension")
plt.ylabel("% of distances within 1 std of the mean")
plt.title("Distance Concentrations")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
