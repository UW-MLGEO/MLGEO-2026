```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist
```

```python
sampleCount = 1000
dimensionCount = [2, 20, 200, 2000]
plt.figure()

for i in dimensionCount:
    data = np.random.rand(sampleCount, i)
    distanceList = pdist(data)
    distanceDistribution = distanceList / np.sqrt(i)
    plt.hist(distanceDistribution, histtype= 'step', bins=100, label=f'{i} dimensions', linewidth=2) 

plt.xlabel('Normalized Distance')
plt.ylabel('Frequency')
plt.legend()
plt.title('Concentration of Distances Across Dimension Counts')
plt.savefig('Justin_CoD_Figure')
plt.show()
```
