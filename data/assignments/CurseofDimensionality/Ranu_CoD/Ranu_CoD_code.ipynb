```python
import numpy as np
import matplotlib.pyplot as plt

n_points = 1000
dims = [1, 2, 5, 10, 50, 100]

ratios = []

for d in dims:
    data = np.random.uniform(0, 1, (n_points, d))
    
    ref = data[0]                     # reference point
    distances = np.linalg.norm(data - ref, axis=1)

    nearest = np.min(distances[1:]) # exclude the reference point itself
    farthest = np.max(distances)

    ratios.append((farthest - nearest) / nearest)

plt.plot(dims, ratios, marker='o')
plt.xlabel("Dimension")
plt.ylabel("(max distance - min distance) / min distance")
plt.title("Curse of Dimensionality: Distance Concentration")
plt.show()
```
