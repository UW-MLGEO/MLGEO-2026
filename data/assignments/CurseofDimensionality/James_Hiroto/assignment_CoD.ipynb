```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

arr1 = np.random.normal(loc=0.0, scale=1.0, size=1000)
arr2 = np.random.normal(loc=0.0, scale=1.0, size=1000)
arr3 = np.random.normal(loc=0.0, scale=1.0, size=1000)
```

```python
arr_3d = np.array([arr1, arr2, arr3]).T
arr_3d
```

```python
arr_3d.shape
```

```python
count_1D = 0
for i in range(len(arr_3d[:,0])):
    if arr_3d[i][0] > 1:
        count_1D += 1
count_1D
```

```python
count_2D = 0
for i in range(len(arr_3d[:,0])):
    if (arr_3d[i][0] > 1) and (arr_3d[i][1] > 1):
        count_2D += 1
count_2D
```

```python
count_3D = 0
for i in range(len(arr_3d[:,0])):
    if (arr_3d[i][0] > 1) and (arr_3d[i][1] > 1) and (arr_3d[i][2] > 1) :
        count_3D += 1
count_3D
```

```python
plt.figure()
plt.plot([count_1D, count_2D, count_3D],'o-')
plt.xticks([0,1,2],["1D","2D","3D"])
plt.grid()
plt.xlabel("Number of dimensions")
plt.ylabel("Number of points above 1")
plt.title("Example of CoD")
```

**Figure 1.** The curse of dimensionality illustrated through data sparsity. Out of 1,000 points sampled from a standard normal distribution, the number of points with all coordinates exceeding 1 drops sharply as dimensions increase—from ~167 in 1D, to ~29 in 2D, to just 5 in 3D. This demonstrates that as dimensionality grows, data become exponentially sparser, making it harder for fixed-size samples to represent the feature space (Altman & Krzywinski, 2018).
