```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Construct a figure illustrating one of the described curses of dimensionality

# As in article, generate 1000 random points in 2D
np.random.seed(21)
points = np.random.randn(1000, 2)
x = points[:,0]
y = points[:,1]
alpha = 1

# Find points within 1 standard deviation of the mean
within_sigma = []
outside_sigma = []
for i in range(len(points)):
    if np.linalg.norm(points[i, :] - alpha*np.array([1, 1])) <= 1:
        within_sigma.append(points[i, :])
    else:
        outside_sigma.append(points[i, :])

within_sigma = np.asarray(within_sigma)
outside_sigma = np.asarray(outside_sigma)

percents = []

for p in np.arange(1,10):
    percents_within = []
    points = np.random.randn(1000, p)
    for i in range(len(points)):
        if np.linalg.norm(points[i, :] - alpha*np.ones((1, p))) <= 1:
            percents_within.append(points[i, :])
    print(len(percents_within))
    percents.append(len(percents_within)/1000)
percents = np.asarray(percents)


fig, axes = plt.subplots(2, 2, figsize = (10,9))
axes[0,0].set_title("Normally Distributed 2D Points, $\mu = 0, \sigma = 1$")
axes[0,0].scatter(within_sigma[:, 0], within_sigma[:, 1], alpha = 0.5, color = "darkblue", label = f"Points within $\sigma$ of ({alpha},{alpha})")
axes[0,0].scatter(outside_sigma[:,0], outside_sigma[:,1], alpha = 0.2, color = 'black', label = f"Points farther than $\sigma$ from ({alpha},{alpha})")
axes[0,0].legend()
axes[0,0].set_xlabel("x")
axes[0,0].set_ylabel("y")
axes[0,0].grid()
axes[0,1].set_title("Distribution of X Values")
axes[0,1].hist(x, alpha = 0.5,color = 'black', bins = 15, label = "2D Data Projected Onto X-Axis")
axes[0,1].hist(within_sigma[:, 0], alpha = 0.5, color = 'darkblue', bins = 5, label = f"2D Data Points Within $\sigma$ of ({alpha}, {alpha})")
axes[0,1].set_xlabel("x")
axes[0,1].set_ylabel("Frequency")
axes[0,1].legend()
axes[0,1].grid()
axes[1,0].set_title("Points Within $\sigma$ of $\mu$ for Increasing Dimensions")
axes[1,0].set_xlabel("Dimension")
axes[1,0].set_ylabel("Percent")
axes[1,0].plot(np.arange(1,10), percents, 'o', linestyle = '-', color = 'darkblue', label = "Percent of Points Within 1 Standard Deviation of the Mean")
axes[1,0].grid()
axes[1,1].set_title("Distribution of Y Values")
axes[1,1].hist(y, alpha = 0.5,color = 'black', bins = 15, label = "2D Data Projected Onto Y-Axis")
axes[1,1].hist(within_sigma[:, 1], alpha = 0.5, color = 'darkblue', bins = 5, label = f"2D Data Points Within $\sigma$ of ({alpha}, {alpha})")
axes[1,1].set_xlabel("x")
axes[1,1].set_ylabel("Frequency")
axes[1,1].legend()
axes[1,1].grid()
fig.tight_layout
fig.suptitle("Explorations of the Curse of Dimensionality")
plt.show()

```
